# Question 3

Count the total number of tokens in the text, including words and punctuation marks. Print the total count.

```python
with open("/content/A Scandal.txt") as file:
  text = file.read()

doc = nlp(text)

words = []

for token in doc:
  if not token.is_punct:
    words.append(token.text)


print("Tota; tokens:",len(doc))  # Using len() to count all tokens,inclusing punctuations
```

# Question 4: Remove Stop Words and Find Lemmas

Remove stop words and punctuation from the text using spaCy. Print the remaining words along with their lemmas.

```python
with  open("/content/A Scandal.txt") as file:
  text = file.read()

doc = nlp(text)

words = []
# Removing stop words and punctuations,then storing the remaining words
for token in doc:
  if not token.is_stop and not token.is_punct :
    words.append(token.lemma_)

print(words)
```

# Question 5: POS Tagging

Use spaCy to identify the part of speech of each token in the text. Print each token along with its POS tag and a short explanation of the tag.

```python
import pandas as pd

with open("/content/A Scandal.txt") as file:
  text = file.read()

doc = nlp(text)

data = []

# Printing each token with its pos tag and description
for token in doc:
  data.append({
      "WORD": token.text,
      "POS"  : token.pos_,
      "Meaning": spacy.explain(token.pos_)
  })

df = pd.DataFrame(data)

print(df)
```

#Question 6: Named Entity Recognition (NER)

Use spaCy to find named entities in the text, such as people, places, and organizations. Print each entity along with its label and the label's explanation.

```python
with open("/content/A Scandal.txt") as file:
  text = file.read()
doc = nlp(text)

data =[]

# extract the named entities and print their labels and decription
for ent in doc.ents:
  data.append({
      "Entity : ent.text"
      "Label": ent.label_,
      "Description" : spacy.explain(ent.label_)
  })

# converting/storing it to datafranme
df = pd.DataFrame(data)
print(df)
```

#Question 7: Find Numbers Using Regular Expressions

Use Python's built-in re module to find all sequences of digits in the text file. Print the numbers you find.

```python
# using a regular expression to find all sequences of digits in the text.

import re

with open("/content/A Scandal.txt") as file:
  text = file.read()

doc = nlp(text)

numbers = re.findall(r"\d+",text)

print(numbers)
```

# Question 8:  Matching

Use spaCy's Matcher to find every occurrence of "singular tragedy" in the text. Print each matching phrase.

```python
# Import the Matcher Library
from spacy.matcher import Matcher
matcher = Matcher(nlp.vocab)

with open("/content/A Scandal.txt") as file:
  text = file.read()

doc = nlp(text)

# Create a pattern to find the phrase in the text
pattern = [{"LOWER":"singular"},
           {"LOWER":"tragedy"}]

matcher.add("singular",[pattern])

matches = matcher(doc)

for match_id,start,end in matches:
    print(doc[start:end].text.replace("\n"," "))



```

#Question 9: Sentence Segmentation

Use spaCy to split the text into individual sentences. Print each sentence separately and count the total number of sentences.

```python
with open("/content/A Scandal.txt") as file:
    text = file.read()

doc = nlp(text)

for sent in doc.sents:
    print(sent.text) # printing sentences
print("-"*40)
print("Number of sentences:", len(list(doc.sents))) #total count of sentences
```
