Q2. Count the total number of words in the PDF.
Print the final word count after processing all pages

```python
# Count word in pdf
from PyPDF2 import PdfReader
reader = PdfReader("/content/GLOBALISATION.pdf")

word_count = 0
for page in reader.pages:
  text = page.extract_text()

  if text:
    word_count += len(text.split())

  print("Total words:",word_count)
```

Q3. Copy the first page of GLOBALISATION.pdf into a new PDF file named first_page.pdf.

```python
# Copy the first page into another pdf

from PyPDF2 import PdfReader,PdfWriter

#Read orignal file
reader = PdfReader("/content/GLOBALISATION.pdf")

#create a new pdf
writer = PdfWriter()

#copy the first page
writer.add_page(reader.pages[0])

#Save as another pdf
with open("first_page.pdf","wb") as output:
  writer.write(output)
```

Q4. Extract the text from every page of the PDF and save it into a text file named pdf_text.txt.
Include a page number heading before each page's text.

```python
# save the extracted text into txt file
from PyPDF2 import PdfReader

reader = PdfReader("/content/GLOBALISATION.pdf")

with open("pdf_text.txt","w",encoding ="utf-8") as file:
  for page_number,page in enumerate(reader.pages,start = 1):

    text = page.extract_text()

    file.write(f"\n--- Page {page_number}---\n")
    file.write(text if text else "")
print("Pdf text is saved")
```

```python
# Import spaCy
import spacy
nlp = spacy.load("en_core_web_sm")

from PyPDF2 import PdfReader

# Open the PDF file in read-binary mode
with open("/content/GLOBALISATION.pdf", "rb") as file:

    # Create a PDF reader object
    reader = PdfReader(file)

    # Create an empty string to store the extracted text
    text = ""

    # Read each page of the PDF
    for page in reader.pages:
        text += page.extract_text() or ""

doc = nlp(text)

for token in doc:
    print(token.text)


```

```python
# Read the PDF file
with open("/content/GLOBALISATION.pdf", "rb") as file:
    reader = PdfReader(file)

    text = ""
    for page in reader.pages:
        text += page.extract_text() or ""

# Process the extracted text
doc = nlp(text)

# Remove stop words and punctuation
for token in doc:
    if not token.is_stop and not token.is_punct:
        # Print the word and its lemma
        print(token.text, "->", token.lemma_)
```

```python
# Read text from the PDF
with open("/content/GLOBALISATION.pdf", "rb") as file:
    reader = PdfReader(file)

    text = ""
    for page in reader.pages:
        text += page.extract_text() or ""

# Process the text
doc = nlp(text)

# Print each token, its POS tag, and explanation
for token in doc:
    print(token.text, token.pos_, spacy.explain(token.pos_))
```

```python
# Read the PDF
with open("/content/GLOBALISATION.pdf", "rb") as file:
    reader = PdfReader(file)

    text = ""
    for page in reader.pages:
        text += page.extract_text() or ""

# Process the text
doc = nlp(text)

# Find and print named entities
for entity in doc.ents:
    print(
        entity.text,
        entity.label_,
        spacy.explain(entity.label_)
    )
```
